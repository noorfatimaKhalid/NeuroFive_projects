```python
import pandas as pd
df=pd.read_csv("titanic.csv")
```

```python
df.isnull().sum()
```

```python
#I choose dropping option as there were only two null values in embark column so it is better to drop them.
#  We can choose filling if number of nulls are greater
df=df.dropna()
```

```python
df.isnull().sum()
```

```python
#Outliers in Age column 
df['Age'].plot(kind='box')
```

```python
#Outliers in Fare column
df['Fare'].plot(kind='box')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.countplot(df, x='Sex',hue="2urvived")
plt.show()
```

```python
sns.barplot(x='Sex',y='Age',data=df)
plt.show()
```

```python
sns.heatmap(df[['Age', 'Fare']].corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
sns.histplot(data=df,x='Sex',y="Age")
plt.show()
```
